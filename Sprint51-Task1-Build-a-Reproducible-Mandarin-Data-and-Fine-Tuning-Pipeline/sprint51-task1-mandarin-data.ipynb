```python
# ============================================================
# Kaggle Script: Download 4 EN-ZH Datasets → finetune_enzh_train.jsonl
# ============================================================

!pip install -q datasets tqdm

from datasets import load_dataset
from tqdm.auto import tqdm
import json
import os

OUTPUT_FILE = "finetune_enzh_train.jsonl"
os.makedirs("/kaggle/working", exist_ok=True)
output_path = f"/kaggle/working/{OUTPUT_FILE}"

# ------------------------------------------------------------
# Helper function to write one record
# ------------------------------------------------------------
def write_record(f, pair_id, source_text, target_text, domain, license_name, dataset_name):
    record = {
        "id": f"{pair_id}_en2zh",
        "pair_id": pair_id,
        "source_text": source_text.strip(),
        "target_text": target_text.strip(),
        "source_lang": "en",
        "target_lang": "zh",
        "domain": domain,
        "license": license_name,
        "dataset_name": dataset_name
    }
    f.write(json.dumps(record, ensure_ascii=False) + "\n")

# ------------------------------------------------------------
# 1. News Commentary
# ------------------------------------------------------------
print("Loading News Commentary...")
news = load_dataset("Helsinki-NLP/news_commentary", "en-zh", split="train")

with open(output_path, "w", encoding="utf-8") as f:
    for i, row in enumerate(tqdm(news, desc="News Commentary")):
        pair_id = f"finetune_pair_{i:06d}"
        write_record(
            f,
            pair_id=pair_id,
            source_text=row["translation"]["en"],
            target_text=row["translation"]["zh"],
            domain="news_commentary",
            license_name="open (WMT/OPUS)",
            dataset_name="Helsinki-NLP/news_commentary"
        )

print(f"News Commentary done. Current file size: {os.path.getsize(output_path)/1e6:.1f} MB")

# ------------------------------------------------------------
# 2. United Nations Parallel Corpus
# ------------------------------------------------------------
print("\nLoading UN Parallel Corpus (this is large)...")
un = load_dataset("Helsinki-NLP/un_pc", "en-zh", split="train")

start_id = len(news)   # continue numbering

with open(output_path, "a", encoding="utf-8") as f:
    for i, row in enumerate(tqdm(un, desc="UNPC")):
        pair_id = f"finetune_pair_{start_id + i:06d}"
        write_record(
            f,
            pair_id=pair_id,
            source_text=row["translation"]["en"],
            target_text=row["translation"]["zh"],
            domain="united_nations",
            license_name="liberal / public domain friendly",
            dataset_name="Helsinki-NLP/un_pc"
        )

print(f"UNPC done. Current file size: {os.path.getsize(output_path)/1e6:.1f} MB")

# ------------------------------------------------------------
# 3. Filtered WikiMatrix
# ------------------------------------------------------------
print("\nLoading Filtered WikiMatrix...")
wiki = load_dataset("larryvrh/WikiMatrix-v1-En_Zh-filtered", split="train")

start_id = start_id + len(un)

with open(output_path, "a", encoding="utf-8") as f:
    for i, row in enumerate(tqdm(wiki, desc="WikiMatrix")):
        pair_id = f"finetune_pair_{start_id + i:06d}"
        write_record(
            f,
            pair_id=pair_id,
            source_text=row["en"],
            target_text=row["zh"],
            domain="wikipedia",
            license_name="CC-BY-SA 4.0",
            dataset_name="larryvrh/WikiMatrix-v1-En_Zh-filtered"
        )

print(f"WikiMatrix done. Current file size: {os.path.getsize(output_path)/1e6:.1f} MB")

# ------------------------------------------------------------
# 4. Tatoeba (eng → cmn)
# ------------------------------------------------------------
print("\nLoading Tatoeba and filtering eng→cmn...")
tatoeba = load_dataset("ymoslem/Tatoeba-Translations", split="train")

en2zh_tatoeba = tatoeba.filter(
    lambda x: x["lang_src"] == "eng" and x["lang_tgt"] == "cmn"
)

start_id = start_id + len(wiki)

with open(output_path, "a", encoding="utf-8") as f:
    for i, row in enumerate(tqdm(en2zh_tatoeba, desc="Tatoeba")):
        pair_id = f"finetune_pair_{start_id + i:06d}"
        write_record(
            f,
            pair_id=pair_id,
            source_text=row["sentence_src"],
            target_text=row["sentence_tgt"],
            domain="conversational",
            license_name="CC-BY 2.0 FR",
            dataset_name="ymoslem/Tatoeba-Translations"
        )

print(f"\nAll done!")
print(f"Final file: {output_path}")
print(f"File size: {os.path.getsize(output_path)/1e6:.1f} MB")
```
